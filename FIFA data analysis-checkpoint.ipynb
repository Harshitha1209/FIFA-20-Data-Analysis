### Checking and Handling the missing values

```python
# Check the missing values
pd.set_option("display.max_rows", None)
df.isnull().sum().sort_values(ascending = False)
```

```python
#Drop the columns which are not important features
cols = ['short_name', 'long_name', 'overall', 'potential',
                 'age', 'nationality', 'club', 'value_eur', 'wage_eur']
df = df.dropna(subset = cols)

# fill the none value in text columns
text_col = ['loaned_from','player_traits','player_tags']
for c in text_col:
    if c in df.columns:
        df[c] = df[c].fillna("None")

# Fill the position columns with the zero
pos_col = ['rs','ls','st','cf','lw','rw','cam','cm','cdm','cb','lb','rb','lwb','rwb','lm','rm','lam','ram','lcm','rcm','ldm','rdm','lcb','rcb','gk','lf','rf','release_clause_eur','contract_valid_until','nation_position',
'nation_jersey_number']
for c in pos_col:
    if c in df.columns:
        df[c] = df[c].fillna(0)

# fill the columns with "-1"
col = ['joined','team_position','team_jersey_number']
for c in col:
    if c in df.columns:
        df[c] = df[c].fillna(-1)

#Fill the GoalKeeper columns with zero
gk_col = [c for c in df.columns if c.startswith('gk_') or c.startswith('goalkeeping_')]
df[gk_col] = df[gk_col].fillna(0)

#Fill the numeric columns with median
numeric_skills = [
    'pace','shooting','passing','dribbling','defending','physic',
    'attacking_crossing','attacking_finishing','attacking_short_passing',
    'attacking_volleys','attacking_heading_accuracy',
    'skill_dribbling','skill_curve','skill_fk_accuracy',
    'skill_long_passing','skill_ball_control',
    'movement_acceleration','movement_sprint_speed','movement_agility',
    'movement_reactions','movement_balance',
    'power_shot_power','power_jumping','power_stamina',
    'power_strength','power_long_shots',
    'mentality_aggression','mentality_interceptions','mentality_positioning',
    'mentality_vision','mentality_penalties','mentality_composure',
    'defending_marking','defending_standing_tackle','defending_sliding_tackle'
]
for c in numeric_skills:
    if c in df.columns:
        df[c] = df[c].fillna(df[c].median())
```

## Determine Number Of Clusters

### Elbow Method

```python
# Elbow method
inertia = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(X_scaled)
    
    inertia.append(kmeans.inertia_)
```

```python
## Visualizing the Elbow method 
plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), inertia, marker="o")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

## Model Training

### KMeans Model

```python
# KMEANS
optimal_k = 4

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

kmeans_clusters = kmeans.fit(X_scaled)
labels = kmeans_clusters.predict(X_scaled)
```

```python
X["KMeans_Cluster"].value_counts().sort_index()
```

### AgglomerativeClustering

```python
from sklearn.cluster import AgglomerativeClustering
agg = AgglomerativeClustering(n_clusters =4,linkage = 'ward')
agg_clusters = agg.fit_predict(X_scaled)
X["Agglomerative_Clusters"] = agg_clusters
X["Agglomerative_Clusters"].value_counts().sort_index()
```

### DBSCAN 

```python
from sklearn.cluster import DBSCAN
dbscan = DBSCAN(eps = 0.8,min_samples = 10)
dbscan_clusters = dbscan.fit_predict(X_scaled)
X["DBSCAN_Clusters"] = dbscan_clusters
X["DBSCAN_Clusters"].value_counts().sort_index()
```

## Clustering Evaluation

### KMEANS

```python

kmeans_score = silhouette_score(X_scaled,labels)
ch_score = calinski_harabasz_score(X_scaled,labels)
db_score = davies_bouldin_score(X_scaled,labels)
print("silhoette score:",kmeans_score)
print("Calinski-Harabasz:",ch_score)
print("Davies-Bouldin Score:",db_score)
```

### Agglomerative

```python
Agg_score = silhouette_score(X_scaled,agg_clusters)
ch_score = calinski_harabasz_score(X_scaled,agg_clusters)
db_score = davies_bouldin_score(X_scaled,agg_clusters)
print("silhoette score:",Agg_score)
print("Calinski-Harabasz:",ch_score)
print("Davies-Bouldin Score:",db_score)
```

### DBSCAN

```python
mask = dbscan_clusters != -1
if len(set(dbscan_clusters[mask])) >= 2:
    score = silhouette_score(X_scaled[mask],dbscan_clusters[mask])
    print("DBSCAN:",score)
    
```

### PCA

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("original:",X_scaled.shape)
print("After Pca:",X_pca.shape)
```

```python
kmeans = KMeans(n_clusters = 4,random_state = 42,n_init = 20)
pca_labels = kmeans.fit_predict(X_pca)
score = silhouette_score(X_pca,pca_labels)
print("PCA + KMeans:",score)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 7))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=pca_labels, cmap="viridis", alpha=0.5, s=8)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Visualization of FIFA 20 Player Clusters")
plt.tight_layout()
plt.savefig("01_pca_clusters.png", dpi=200, bbox_inches="tight")
plt.show()
```
